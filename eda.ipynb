```python
stat_2014 = pd.DataFrame(
    {
        "x1": [0.49, 0.48, 0.20, 0.63, 0.33, 0.30],
        "x2": [0.035, 0.031, 0.472, 0.073, 0.018, 0.055],
        "x3": [0.44, 0.45, 0.25, 0.62, 0.23, 0.39],
        "x4": [18, 12, 0.73, 50, 5, 45],
        "x5": [3.8, 3.3, 0.50, 8.9, 1.6, 7.3],
        "x6": [78, 75, 0.22, 119, 55, 64],
        "x7": [1291, 1305, 0.24, 1880, 801, 1078],
        "x8": [5.8, 6.3, 0.23, 7.9, 2.9, 5.0],
    },
    index=["Średnia", "Mediana", "Wsp. zmienności", "Max", "Min", "Rozstęp"],
)

skumulowany_CPI_2007_14 = 1.2378
stat_2014.loc[["Średnia", "Mediana", "Max", "Min", "Rozstęp"], ["x5", "x6", "x8"]] *= skumulowany_CPI_2007_14
stat_2014 = stat_2014.round(2)
stat_2014.loc[["Średnia", "Mediana", "Max", "Min", "Rozstęp"], ["x1", "x2", "x3"]] *= 100

stat_2023 = eda.stats(df)

skumulowany_CPI_2014_23 = 1.4463
stat_2023.loc[["Średnia", "Mediana", "Max", "Min", "Rozstęp"], ["x5", "x6", "x8"]] /= skumulowany_CPI_2014_23
stat_2023 = stat_2023.round(2)


eda.df_compare(stat_2014, stat_2023, title1="2014", title2="2023")
```
